# Building Code Agent with Local Transformer Qwen2.5-0.5B-Instruct Model with SmolaAgents

## Overview
This notebook demonstrates how to build and use a **Code Agent** powered by the SmolaAgents framework with a local transformer model.

## What is SmolaAgents?
SmolaAgents is a lightweight, open-source framework for building AI agents that can:
- Reason through problems step-by-step
- Write and execute code to solve tasks
- Interact with various tools and APIs
- Run efficiently on local machines

## Key Components
- **Model**: Qwen2.5-0.5B-Instruct (a small, lightweight language model)
- **Framework**: SmolaAgents with TransformersModel backend
- **Agent Type**: CodeAgent (can generate and execute Python code)

## What You'll Learn
1. How to load a language model using Hugging Face Transformers
2. How to initialize a CodeAgent with SmolaAgents
3. How to prompt an agent to solve problems (e.g., mathematical calculations)
4. How to capture and display agent outputs

Let's get started!

```python
import warnings
warnings.filterwarnings("ignore")
```

## Example 1: Addition

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

model_id = "Qwen/Qwen2.5-0.5B-Instruct"

from transformers import AutoModelForCausalLM, AutoTokenizer
from smolagents import CodeAgent, TransformersModel

agent = CodeAgent(model=TransformersModel(
    model_id,
    temperature=0.7,
    max_new_tokens=1000), tools=[])

result = agent.run("Calculate the sum of numbers from 1 to 100")
print(result)
```

## Example 2: Data Structure Manipulation

Let's ask the agent to work with more complex data structures - filtering, sorting, and transforming data.

```python
# Example 2: Data manipulation
task2 = """
Given a list of students with their scores:
students = [{'name': 'Alice', 'score': 85}, {'name': 'Bob', 'score': 92}, 
            {'name': 'Charlie', 'score': 78}, {'name': 'Diana', 'score': 95}]

Find the top 2 students by score and return their names and scores.
"""

result2 = agent.run(task2)
print("Task 2 Result:")
print(result2)
```

## Example 3: String Processing & Text Analysis

Here the agent performs text analysis - counting words, finding patterns, and analyzing string properties.

```python
# Example 3: Text analysis
task3 = """
Analyze this text and provide:
1. Total word count
2. Number of unique words
3. Average word length
4. Most common word (excluding articles like 'the', 'a', 'an')

Text: "The quick brown fox jumps over the lazy dog. The quick fox is clever and nimble. 
      The dog is lazy but loyal. Fox and dog are common in stories."
"""

result3 = agent.run(task3)
print("Task 3 Result:")
print(result3)
```

## Example 4: Algorithmic Problem Solving

The agent solves a more complex algorithmic problem - this tests logical reasoning and algorithm implementation.

```python
# Example 4: Algorithmic problem
task4 = """
Write a function to find all prime numbers up to 50.
Return them as a list and also calculate:
1. How many primes are there?
2. What is the sum of all primes?
3. What is the average of all primes?
"""

result4 = agent.run(task4)
print("Task 4 Result:")
print(result4)
```

## Example 5: Complex Data Processing Pipeline

A comprehensive example combining multiple operations - this demonstrates the agent's ability to handle multi-step reasoning and complex workflows.

```python
# Example 5: Complex data pipeline
task5 = """
Process this dataset and create a summary report:

sales_data = [
    {'month': 'Jan', 'product': 'A', 'quantity': 100, 'price': 50},
    {'month': 'Jan', 'product': 'B', 'quantity': 150, 'price': 30},
    {'month': 'Feb', 'product': 'A', 'quantity': 120, 'price': 50},
    {'month': 'Feb', 'product': 'B', 'quantity': 100, 'price': 30},
    {'month': 'Mar', 'product': 'A', 'quantity': 200, 'price': 50},
    {'month': 'Mar', 'product': 'B', 'quantity': 180, 'price': 30}
]

Create a report with:
1. Total revenue by product
2. Total revenue by month
3. Best performing product (by revenue)
4. Which month had the highest revenue
"""

result5 = agent.run(task5)
print("Task 5 Result:")
print(result5)
```

## Key Observations

### What Worked Well
- **Simple mathematical operations**: The agent excels at basic calculations and arithmetic
- **Data structure manipulation**: Lists and dictionaries are handled effectively
- **Logical reasoning**: The agent can break down complex tasks into steps

### Challenges & Limitations
- **Small model constraints**: Qwen2.5-0.5B has limited reasoning capacity
- **Context length**: Complex multi-step problems may exceed token limits
- **Hallucinations**: The agent may generate plausible-sounding but incorrect code
- **Complex algorithms**: Very intricate algorithms may not be correctly implemented

### Tips for Better Results
1. **Break down complex tasks**: Split large problems into smaller sub-problems
2. **Provide clear examples**: Include expected input/output formats
3. **Use explicit prompting**: Be specific about what you want
4. **Validate outputs**: Always check agent-generated code before using it in production
5. **Use larger models**: For complex reasoning, consider using larger models like Llama-2 or GPT-4

### When to Use Code Agents
✅ Good for:
- Data processing pipelines
- Prototyping code solutions
- Automating routine programming tasks
- Educational purposes

❌ Not ideal for:
- Mission-critical systems
- Complex mathematical proofs
- Security-sensitive code
- Production systems (without validation)
