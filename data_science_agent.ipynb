```python
from transformers import AutoModelForCausalLM, AutoTokenizer

model_id = "Qwen/Qwen2.5-0.5B-Instruct"

from transformers import AutoModelForCausalLM, AutoTokenizer
from smolagents import CodeAgent, TransformersModel, Tool
from smolagents.tools import DuckDuckGoSearch, WebSearchTool, WikipediaSearchTool

agent = CodeAgent(model=TransformersModel(
    model_id,
    temperature=0.7,
    max_new_tokens=5000), tools=[
        DuckDuckGoSearch(),
        WebSearchTool(),
        WikipediaSearchTool()], additional_authorized_imports=["pandas", "numpy", "matplotlib", "sklearn"])

task = "Calculate the sum of numbers from 1 to 100"
result = agent.run(task)
print(result)
```
